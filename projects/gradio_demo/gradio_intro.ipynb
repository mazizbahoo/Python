```python
def greet(name):
    return "Welcome " + name + "!"

gradio_interface = gr.Interface(fn=greet, inputs="text", outputs="text")
gradio_interface.launch()
```

```python
def greet(message):
    return "Welcome " + message + "!"
    
gradio_interface = gr.Interface(fn=greet, inputs=gr.Textbox(lines=4, placeholder="Type your message here..."), outputs="text")
gradio_interface.launch()
```

```python
def interesting(name, is_morning, temperature):
    salutation = "Good Morning" if is_morning else "Good Evening"
    celcius = (temperature-32)*5/9
    greeting = f"{salutation} {name}. It is {temperature} farenheit today."
    return greeting, celcius

gradio_interface = gr.Interface(fn=interesting, inputs=["text", "checkbox", gr.Slider(1, 100)], outputs=["text", "number"])
gradio_interface.launch()
```

```python
# Gradio Block

def block(name):
    return "Welcome " + name + "!"
with gr.Blocks() as demo:
    name = gr.Textbox(label="Name")
    output = gr.Textbox(label="Output Box")
    block_btn = gr.Button("Greeting")
    block_btn.click(fn=block, inputs=name, outputs=output)

demo.launch()
```
