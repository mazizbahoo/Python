```python
import pickle
import gradio as gr
import pandas as pd
```

```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
```

```python
X, y = load_iris(return_X_y=True)
```

```python
clf = LogisticRegression(random_state=0).fit(X, y)
```

```python
with open("model_iris.pkl", "wb") as f:
    pickle.dump(clf, f)
```

```python
def make_prediction(sepal_length, sepal_width, petal_length, petal_width):
    with open("model_iris.pkl", "rb") as f:
        clf = pickle.load(f)
        prediction = clf.predict([[sepal_length, sepal_width, petal_length, petal_width]])
        if prediction == 0:
            return "Iris-Setosa"
        elif prediction == 1:
            return "Iris-Versicolor"
        else:
            return "Iris-Virginica"
```

```python
# Create the input component for gradio since we are expecting four inputs.
sl = gr.Number(label="Sepal Length")
sw = gr.Number(label="Sepal Width")
pl = gr.Number(label="Petal Length")
pw = gr.Number(label="Petal Width")
```

```python
output = gr.Textbox(label="Predicted Iris Species")
```

```python
app = gr.Interface(fn=make_prediction, inputs=[sl, sw, pl, pw], outputs=output)
app.launch()
```
