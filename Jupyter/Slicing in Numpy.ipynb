```python
# Array Arithmetics:
print("Arrays Arithmetics")
arr=np.arange(1,11)
print(arr)
arr1=np.arange(21,31)
print(arr1)
```

```python
arr2 = arr1 + arr 
print("Add two arrays", arr2)
arr2 = arr1 - arr 
print("Subtract two arrays", arr2)
arr3 = arr1 * arr
print("Mutiplication of two arrays", arr3)
```

```python
# Addition, Subtraction, Multiplication, Division, √SquareRoot, and Square of two arrays generated above.
arr_square = arr * arr
print("Sqaure of Numpy Arr:", arr_square)
```

```python
print("√SquareRoot of Numpy Arr:", np.round(np.sqrt(arr), 2))
```

```python
# ndenumerate
for idx, values in np.ndenumerate(arr):
    if values==5:
        print(f' Index of {values} is {idx}')
```

```python
for i in range(arr.size):
    if arr[i]==9:
        print(f'Index of {arr[i] is [i]}')
```

```python
import numpy as np
```

```python
#check there exist zero?
x=np.array([1,2,3])
print("Original Array",x)
print("Test if none of the elements of the array is zero")
print(np.all(x)) 
```

```python
#Test wheather any of the elements of given arrays is non-zero.
s=np.array([0,0,0,0])
print("Original Array:",s)
print("Test wheather any of elements of given array is non-zero")
print(np.any(s))
```

```python
#Infinite number:
x=np.array([1,0,np.inf,0])
print("Original Array:",x)
print("Test wheather any of element of given array is infinte")
print(np.isinf(x))
```

```python
#nan= not a number 
x=np.array([1,0,np.nan,0])
print("Original Array:",x)
print("Test wheather any of element of given array is nan")
print(np.isnan(x))
```

```python
#complex and real + imaginary part:
x=np.array([1+2j,1+0j,4.5,3,2,2j])
print("Original Array:",x)
print("Test wheather any of element of given array is complex number")
print(np.iscomplex(x))
```
