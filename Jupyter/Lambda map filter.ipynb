```python
b=10
x=lambda a:a+b
y=x(5)
print(y)
```

```python
x=lambda a,b :a*b
y=x(5,6)
print(y)
```

```python
x= lambda a,b,c: a+b+c
y=x(4,5,6)
print(y)
```

```python
def myfun(a):
    return len(a)


x=map(myfun,('apple','banana','cherry'))
print('map fun output',list(x))
```

```python
def myfun1(a,b):
    return a,b


x=map(myfun1,('apple','banana','cherry'),('orange','watermelon','grapes'))
print('map fun output',list(x))
```

```python
#Filter function
def myfun3(x):
    if x<18:
        return False
    else:
        return True



ages=[5,12,17,18,26,32] 
adults=filter(myfun3,ages)
print('Filter function output', list(adults))
```

```python
#Can we use map in lamba
number=[1,2,3,4,5]
sq_num=list(map(lambda x: x**2 , number))
print(sq_num)
```
